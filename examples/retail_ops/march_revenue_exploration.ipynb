# Retail ops exploration

First pass on order revenue. Stakeholders asked about March performance.

```python
import pandas as pd
orders = pd.read_csv('orders.csv', parse_dates=['order_date'])
customers = pd.read_csv('customers.csv')
products = pd.read_csv('products.csv')
print(f'orders={len(orders)}')
```

```python
orders['month'] = orders['order_date'].dt.to_period('M').astype(str)
monthly = orders.groupby('month')['revenue'].sum()
monthly
```

## Changed direction

Monthly totals alone are ambiguous. Switching to region × category after joining customers/products.

```python
df = orders.merge(customers, on='customer_id', how='left').merge(products, on='product_id', how='left')
print('joined preview ready')
df.head()
```

```python
# Intentionally fragile: assumes join already happened
df[df['month'] == '2024-03'].groupby(['region', 'category'])['revenue'].sum()
```
