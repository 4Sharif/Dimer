# Sales exploration

Initial look at regional revenue.

```python
import pandas as pd
df = pd.read_csv('examples/sales/sales.csv')
print(f'loaded {len(df)} rows')
```

```python
by_region = df.groupby('region')['revenue'].sum().sort_values(ascending=False)
by_region.head()
```

## Changed direction

Instead of regional totals, pivot to monthly trends because stakeholders asked about March.

```python
df['month'] = pd.to_datetime(df['date']).dt.to_period('M').astype(str)
monthly = df.groupby('month')['revenue'].sum().sort_values(ascending=False)
monthly.head()
```

```python
missing_pct = df['revenue'].isna().mean()
missing_pct
```
